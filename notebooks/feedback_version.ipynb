## Translation with Feedback Loop
This notebook demonstrates the feedback-based version of the KNIME2Galaxy translation pipeline.
In this approach, the pipeline is extended with an iterative feedback loop: if an error occurs during execution, the error information is sent back to the LLM. 
The model then re-evaluates the generated `.ga` workflow and attempts to fix the identified issues.
The updated workflow is validated again, and this process repeats until a valid and executable Galaxy workflow is produced OR a maximum number of iterations is reached.
The overall goal is to reliably translate a KNIME workflow into a `.ga` file that can be executed in Galaxy without errors.

```python
from imaging_knime_to_galaxy.feedback import translate_feedback
from pathlib import Path
```

```python
data_folder = Path("../data").resolve()
base_folder = Path("..").resolve()
```

### First, set all paths and file names

```python
knwf_path = data_folder / "file_to_translate" / "2025_03_2D_spot_detection.knwf"

tools_metadata_path = data_folder / "tools_metadata.json"
translation_table_path = data_folder / "translation_table.yml"
workflow_examples_yml_path = data_folder / "workflow_translation_table.yml"
output_galaxy_workflow_path = data_folder / "output_file" / f"FEEDBACK_2025_03_2D_spot_detection.ga"
output_galaxy_workflow_path_labeled = data_folder / "output_file" / f"FEEDBACK_2025_03_2D_spot_detection_labeled.ga"
input_workflow_path = data_folder / "input_workflows.ga"
vector_store_path = data_folder / "vector_store.npz"
input_image_path = base_folder / "knime2galaxy_scheme.png"    # example image will be the pipeline scheme from the ReadMe
job_yml_path = data_folder / "job.yml"
```

### Then call the translation pipeline with feedback mechanism

```python
record = translate_feedback(
    knwf_path,tools_metadata_path,translation_table_path,
    workflow_examples_yml_path,output_galaxy_workflow_path,
    output_galaxy_workflow_path_labeled,input_workflow_path,vector_store_path,
    input_image_path,job_yml_path,max_iterations=5)
```
